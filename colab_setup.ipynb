## 3. Set your Roboflow API key

Paste your key below. Get it from app.roboflow.com → Account → API Keys.

````python
import os

os.environ["ROBOFLOW_API_KEY"]  = "YOUR_API_KEY_HERE"   # ← paste your key
os.environ["ROBOFLOW_WORKSPACE"] = "evan-lee-rrndd"
os.environ["ROBOFLOW_PROJECT"]   = "acne04-detection-p8j0d"
os.environ["ROBOFLOW_VERSION"]   = "1"

# Write a .env file so the loader script can also read it
with open("/content/AcneDetection/.env", "w") as f:
    f.write(f"ROBOFLOW_API_KEY={os.environ['ROBOFLOW_API_KEY']}\n")
    f.write(f"ROBOFLOW_WORKSPACE={os.environ['ROBOFLOW_WORKSPACE']}\n")
    f.write(f"ROBOFLOW_PROJECT={os.environ['ROBOFLOW_PROJECT']}\n")
    f.write(f"ROBOFLOW_VERSION={os.environ['ROBOFLOW_VERSION']}\n")

print("Environment variables set.")
````
